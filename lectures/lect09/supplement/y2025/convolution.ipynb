### Edge detection with convolution

```python
x = np.array([[4,2,9,4,7,1],
             [3,1,5,3,2,5],
             [8,7,1,2,6,2],
             [6,3,0,1,8,1],
             [1,4,1,5,7,8],
             [2,0,9,1,3,9]])
h = np.array([[1,0,-1],
             [1,0,-1],
             [1,0,-1]])
h
```

```python
h = h[::-1,::-1]
```

```python
h
```

```python
signal.convolve2d(x,h,mode='valid')
```

```python
# small image data with a vertical edge
x1 = np.array([[20,20,20,0,0,0],
             [20,20,20,0,0,0],
             [20,20,20,0,0,0],
             [20,20,20,0,0,0],
             [20,20,20,0,0,0],
             [20,20,20,0,0,0]])
x1
```

```python
signal.convolve2d(x1,h,mode='valid')
```

#### Example 1 : vertical edge detection

```python
ascent = scipy.datasets.ascent()
plt.imshow(ascent, cmap='gray')
plt.axis('off')
plt.show()
```

```python
ascent_vertedge = signal.convolve2d(ascent,h,mode='same')
plt.imshow(ascent_vertedge, cmap='gray',vmin=0,vmax=255)
plt.axis('off')
plt.show()
```
